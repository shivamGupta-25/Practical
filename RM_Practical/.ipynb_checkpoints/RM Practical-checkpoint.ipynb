#### Spearman's Rank Correlation (p)

```python
from scipy.stats import spearmanr

math_ranks = [1, 2, 3, 4, 5]
science_ranks = [2, 1, 4, 3, 5]

rho, p_value = spearmanr(math_ranks, science_ranks)

print("Spearman's Rank Correlation (p):", rho)
print("P-Value:", p_value)
```

#### Chi-square goodness-of-fit test (x2)

```python
from scipy.stats import chisquare

observed = [8, 9, 10, 12, 11, 10]
expected = [10, 10, 10, 10, 10, 10]

chi2, p_value = chisquare(f_obs = observed, f_exp = expected)

print("Chi-square goodness-of-fit (x2):",chi2)
print("P-Value:", p_value)
```

#### ANOVA

```python
from scipy.stats import f_oneway

method_a = [85, 90, 88, 75, 95] 
method_b = [78, 74, 80, 76, 82] 
method_c = [82, 88, 85, 90, 87]

f, p_value = f_oneway(method_a, method_b, method_c)

print("Anova (f):", f)
print("P-Value:", p_value)
```

#### T-Test

```python
from scipy.stats import ttest_ind

Group_A = [78, 85, 92, 88, 75] 
Group_B = [70, 65, 80, 72, 68]

ttest, p_value = ttest_ind(Group_A, Group_B)
print("T-test:", ttest)
print("P-Value:", p_value)
```

#### Paired T-Test

```python
from scipy.stats import ttest_rel

before = [70, 68, 80, 72, 65, 74] 
after  = [75, 74, 85, 78, 70, 79] 

ttest_paired, p_value = ttest_rel(after, before)

print("Paired T-test:", ttest_paired)
print("P-Value:", p_value)
```

#### Pearson correlation coefficient (r)

```python
from scipy.stats import pearsonr

x = [2, 3, 5, 6, 8]
y = [65, 70, 75, 78, 85]

r, p_value = pearsonr(x, y)

print("Pearson correlation coefficient (r):", r)
print("P-Value:", p_value)
```

#### Coefficient of Determination (R2)

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Data
X = np.array([2, 3, 5, 6, 8]).reshape(-1, 1)
Y = np.array([65, 70, 75, 78, 85])

# Model
model = LinearRegression()
model.fit(X, Y)

# R² score
r_squared = model.score(X, Y)
print(f"Coefficient of Determination (R²): {r_squared:.4f}")
```
